```python
# Import libraries for data analysis
import pandas as pd
import numpy as np

# Load datasets from the parent folder's data directory
movies = pd.read_csv("../data/tmdb_5000_movies.csv")
credits = pd.read_csv("../data/tmdb_5000_credits.csv")

# Display dataset dimensions
print("Movies dataset:", movies.shape)
print("Credits dataset:", credits.shape)
```

```python
# Display all column names in the movies dataset
movies.columns.tolist()
```

```python
# Display all column names in the credits dataset
credits.columns.tolist()
```

```python
# Display selected columns to understand the movie information
display(movies[['title', 'genres', 'keywords', 'overview']].head())

# Display the first five credits records
display(credits.head())
```

```python
# Count missing values in both datasets
print("Missing values in movies:")
display(movies.isnull().sum()[movies.isnull().sum() > 0])

print("Missing values in credits:")
display(credits.isnull().sum()[credits.isnull().sum() > 0])
```

```python
# Count duplicate rows in both datasets
print("Duplicate movie rows:", movies.duplicated().sum())
print("Duplicate credit rows:", credits.duplicated().sum())
```

```python
# Merge movie details with their corresponding cast and crew data
movies = movies.merge(
    credits,
    left_on="id",
    right_on="movie_id",
    suffixes=("", "_credits")
)

# Check the merged dataset
print("Merged dataset shape:", movies.shape)
display(movies[["id", "title", "genres", "keywords", "overview", "cast", "crew"]].head())
```

```python
# Check missing values in the features we may use for recommendations
columns_to_check = ["title", "genres", "keywords", "overview", "cast", "crew"]

display(movies[columns_to_check].isnull().sum())
```

```python
# Replace missing overviews with empty strings
movies["overview"] = movies["overview"].fillna("")

# Verify missing values in the important columns
columns_to_check = ["title", "genres", "keywords", "overview", "cast", "crew"]

print(movies[columns_to_check].isnull().sum())
```

```python
# Display one genres value to understand its structure
print(movies.loc[0, "genres"])

# Display one keywords value
print(movies.loc[0, "keywords"])

# Display one cast value
print(movies.loc[0, "cast"])
```

```python
# Import a library to safely convert string representations into Python objects
import ast
```

```python
# Extract the name values from a JSON-like string
def extract_names(text):
    items = ast.literal_eval(text)
    return [item["name"] for item in items]
```

```python
# Test extraction on the first movie's genres, keywords, and cast
print("Genres:", extract_names(movies.loc[0, "genres"]))
print("Keywords:", extract_names(movies.loc[0, "keywords"]))
print("Cast:", extract_names(movies.loc[0, "cast"])[:5])
```

```python
# Convert genres, keywords, and cast strings into lists of names
movies["genres"] = movies["genres"].apply(extract_names)
movies["keywords"] = movies["keywords"].apply(extract_names)
movies["cast"] = movies["cast"].apply(extract_names)

# Keep only the first three cast members for each movie
movies["cast"] = movies["cast"].apply(lambda actors: actors[:3])

# Display the extracted features
display(movies[["title", "genres", "keywords", "cast"]].head())
```

```python
# Extract the director's name from the crew information
def extract_director(text):
    crew_list = ast.literal_eval(text)

    for person in crew_list:
        if person["job"] == "Director":
            return person["name"]

    return ""

# Apply the function to every movie
movies["director"] = movies["crew"].apply(extract_director)

# Check the extracted directors
display(movies[["title", "director"]].head(10))
```

```python
# Check whether any feature extraction produced missing values
print("Missing genres:", movies["genres"].isnull().sum())
print("Missing keywords:", movies["keywords"].isnull().sum())
print("Missing cast:", movies["cast"].isnull().sum())
print("Missing directors:", movies["director"].isnull().sum())
```

```python
# Normalize multiword names and keywords into single tokens
def normalize_names(items):
    return [item.replace(" ", "").lower() for item in items]

# Apply normalization to list-based features
movies["genres"] = movies["genres"].apply(normalize_names)
movies["keywords"] = movies["keywords"].apply(normalize_names)
movies["cast"] = movies["cast"].apply(normalize_names)

# Normalize the director name
movies["director"] = movies["director"].apply(
    lambda name: name.replace(" ", "").lower()
)

# Convert overview text to lowercase words
movies["overview"] = movies["overview"].str.lower()

# Display the normalized features
display(movies[["title", "genres", "keywords", "cast", "director", "overview"]].head(3))
```

```python
# Combine the selected features into one text field per movie
movies["tags"] = (
    movies["overview"] + " " +
    movies["genres"].apply(lambda x: " ".join(x)) + " " +
    movies["keywords"].apply(lambda x: " ".join(x)) + " " +
    movies["cast"].apply(lambda x: " ".join(x)) + " " +
    movies["director"]
)

# Keep only the columns needed for the recommendation system
movies_clean = movies[["id", "title", "tags"]].copy()

# Inspect the combined tags
display(movies_clean.head())

# Check the final dataset dimensions
print("Clean dataset shape:", movies_clean.shape)
```

```python
# Import TF-IDF to convert movie tags into numerical vectors
from sklearn.feature_extraction.text import TfidfVectorizer

# Create the TF-IDF vectorizer
tfidf = TfidfVectorizer(max_features=5000, stop_words="english")

# Convert the combined tags into a TF-IDF matrix
tfidf_matrix = tfidf.fit_transform(movies_clean["tags"])

# Display the matrix dimensions
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
# Import cosine similarity
from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity scores between all movies
similarity = cosine_similarity(tfidf_matrix)

# Display the similarity matrix dimensions
print("Similarity matrix shape:", similarity.shape)

# Check Avatar's similarity with itself
print("Avatar similarity with itself:", similarity[0][0])
```

```python
# Create a function that recommends similar movies
def recommend_movies(movie_title, number_of_recommendations=5):
    # Find the selected movie
    matching_movies = movies_clean[
        movies_clean["title"].str.lower() == movie_title.lower()
    ]

    if matching_movies.empty:
        return f"Movie '{movie_title}' not found in the dataset."

    # Get the index of the selected movie
    movie_index = matching_movies.index[0]

    # Pair each movie index with its similarity score
    scores = list(enumerate(similarity[movie_index]))

    # Sort by similarity score in descending order
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    # Exclude the selected movie and return the top recommendations
    recommended_indices = [
        index for index, score in scores if index != movie_index
    ][:number_of_recommendations]

    return movies_clean.iloc[recommended_indices][["title"]]
```

```python
# Recommend five movies similar to Avatar
recommend_movies("Avatar")
```

```python
# Recommend five movies similar to The Dark Knight Rises
recommend_movies("The Dark Knight Rises")
```

```python
# Import libraries for file and directory handling
from pathlib import Path
import joblib

# Create the artifacts directory in the project root
ARTIFACTS_DIR = Path("../artifacts")
ARTIFACTS_DIR.mkdir(parents=True, exist_ok=True)

print("Artifacts directory:", ARTIFACTS_DIR.resolve())
```

```python
# Save movie IDs, titles, and combined tags
movies_clean.to_pickle(ARTIFACTS_DIR / "movies_clean.pkl")

print("Cleaned movie data saved successfully.")
```

```python
# Save the precomputed cosine similarity matrix
joblib.dump(similarity, ARTIFACTS_DIR / "similarity.pkl", compress=3)

print("Similarity matrix saved successfully.")
```

```python
# Check that both artifact files exist and display their sizes
for file_path in ARTIFACTS_DIR.iterdir():
    size_mb = file_path.stat().st_size / (1024 * 1024)
    print(f"{file_path.name}: {size_mb:.2f} MB")
```
